## Basic Definitions

In object-oriented programming such as Python, a **class**  provides a means of bundling data and functionality together. A class can be thought of a template for defining objects that share the same methods and properties. In other words,  a class is like an object constructor for creating these objects. In this context, an **object** is a specific **instance** of a class. 

* The Python naming convention for classes follows the [camel case](https://en.wikipedia.org/wiki/Camel_case), where each word starts with a capital letter and separate words are not separated (e.g., `MyFirstClass`).
* All classes have a function called `__init__()`, which is executed automatically when the class is being used to create a new instance. The `__init__()` function is used to perform operations (e.g., assigning values to object properties) when a class instance is being created. 
* An **instance attribute** is a variable that belongs to the instance of a class. It is defined in the constructor function `__init__()`.
* A **class attribute** is a variable that belongs to the class itself and that is shared across all instances. It is defined outside the constructor function `__init__()`. 

```python
class Circle:
    """Circle class"""
    pi = 3.14159
    def __init__(self, r=1):
        self.radius = r
    def area(self):
        """Compute area"""
        area = self.radius * self.radius * Circle.pi
        return area
    
print(f"Class docstring: {Circle.__doc__}\n")

c1 = Circle(3)
c2 = Circle(1)
print(f"c1.pi = {c1.pi},  c2.pi = {c2.pi}")
print(f"Area for r={c1.radius}: {c1.area()}, Area for r={c2.radius}: {c1.area()}\n")

Circle.pi = 3
print(f"c1.pi = {c1.pi},  c2.pi = {c2.pi}")
print(f"Area for r={c1.radius}: {c1.area()}, Area for r={c2.radius}: {c1.area()}\n")

c1.pi = 2
print(f"c1.pi = {c1.pi},  c2.pi = {c2.pi}")
print(f"Area for r={c1.radius}: {c1.area()}, Area for r={c2.radius}: {c1.area()}")
print(f"Note that Circle.pi is used to compute the area and not c1.pi")
```

A class can have different types of attributes. For example, in the prior example, the class `Circle` has the attribute `Circle.pi` returning an integer, the attribute `Circle.area` returning a function object, and the attribute `Circle.__doc__` returning the docstring `Circle class`.

The `self` parameter refers to the current class instance and allows for accessing variables that belong to the class. The parameter may be given a different name, but it has to be the first parameter of any function of the class. In other words, the name `self` has absolutely no special meaning to Python, but should be used for the sake of readability of the code.

```python
class Circle:
    """Circle class"""
    pi = 3.14159                              # class variable shared by all instances
    def __init__(self_different_name, r=1):
        self_different_name.radius = r        # instance variable unique to each instance
    def area(self_another_name):
        """Compute area"""
        area = self_another_name.radius * self_another_name.radius * Circle.pi
        return area
    
c = Circle(1)
print(c.area())  
```

Note that, for the class `Circle`, `Circle.area` is a **function object**. However, for an instance `c = Circle(1)`, `c.area` is not a function object but a **method object**. Note that `Circle.area` has an argument (namely, `self`), while `c.area` has no argument. In other words, `Circle.area` is not the same thing as `c.area`. Actually, the instance object is passed as the first argument of the function, making the two following calls of the function equivalent: 

```python
c = Circle(1)
print(f"Call for method object: {c.area()}")
print(f"Call for function object: {Circle.area(c)}")
```

One can modify instance or class properties:

```python
c = Circle(3)
Circle.pi = 3.14159
c.radius = 3
print(c.area())

c.radius = 1
print(c.area())

Circle.pi = 3
print(c.area())
```

One can delete object properties or entire objects:

```python
c = Circle(3)
print(c.__dict__)
del c.radius
print(c.__dict__)
del c
```

The `pass` statement is a null statement which can be used as a placeholder for future code. It can be used to define a class without any content.

```python
class Nothing:
    pass

a = Nothing()
print(a.__dict__)
```

## Standard Attributes of Classes

The `__dict__` attribute, which is available to all Python objects, contains all the attributes that describe the object or class under question.

```python
print(f"__dict__ for class\n {Circle.__dict__}\n")
c = Circle(3)
print(f"__dict__ for class instance\n {c.__dict__}\n")
```

The `__doc__` attribute provides the documentation string of the class, function, or attribute.

```python
print(Circle.__doc__)
print(c.__doc__)
print(Circle.area.__doc__)
print(c.area.__doc__)
```

The `__name__` attribute contains the class name.

```python
print(Circle.__name__)
```

The `__class__` attribute refers to the class from which the object was created. Note that each value is an object, and therefore has a **class** (also called its **type**). 

```python
print(Circle.__class__)
print(c.__class__)
value = 10
print(value.__class__)
value = 1.1
print(value.__class__)
```

## Inheritance

In object-oriented programming, **class inheritance** is a concept for defining a class (called **child class** or **derived class**) that inherits all the methods and properties from another class (called **parent class** or **base class**). When creating a child class, the parent class is used as a parameter:

```python
class Person():
    currentyear = 2023
    def __init__(self, firstname, lastname, birthyear):
        self.firstname = firstname
        self.lastname = lastname
        self.birthyear = birthyear
    def age(self):
        return Person.currentyear - self.birthyear
    def print_info(self):
        print(f"{self.firstname} {self.lastname} is born in the year {self.birthyear}.")
    
class Student(Person):
    pass
    
p = Person("Michael", "Smith", 1953)
p.print_info()
print(f"Age in the year {Person.currentyear}: {p.age()}")

s = Student("David", "Wilson", 2003)
s.print_info()
print(f"Age in years {Student.currentyear}: {s.age()}")
```

Note that a child class can override any methods of its parent class. For example, simply adding the `__init__()` function to the child class will override the inheritance of the parent's `__init__()` function. To avoid this, the `super()` function can be used to make the child inherit all the methods and properties from its parent.

```python
class Student(Person):
    def __init__(self, firstname, lastname, birthyear, university):
        super().__init__(firstname, lastname, birthyear)
        self.university = university
    def print_info(self):
        super().print_info()
        print(f"{self.firstname} ({self.age()} years) studies at {self.university}.")
        
s = Student("David", "Wilson", 2003, "NYU")
s.print_info()
```

<a id='notes'></a> 
## Further Notes

*  Python allows for multiple inheritance: 
  `Class A: ...`, `Class B: ...`, `Class C(A,B): ...`<br> 
  In this case, class `C`inherits all variables and methods from `A` as well as `B`. In situations where classes `A` and `B` share method names, Python uses certain order conventions to choose which method to be used. In practice, it is advised to avoid name clashes.
