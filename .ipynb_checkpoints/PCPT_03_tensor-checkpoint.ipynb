<h2>Overview and Learning Objectives</h2>

This unit is dedicated to exploring PyTorch tensors, a fundamental data structure in PyTorch that closely resembles arrays and matrices. Tensors are crucial for representing data, model inputs and outputs, and neural network parameters within PyTorch. This unit covers various aspects of tensors, including their creation, attributes, data types, operations, storage, and serialization. The learning objectives of this unit are as follows:

<ul>    
<li> Learn how to create PyTorch tensors through functions like torch.zeros, torch.ones, and torch.rand or by converting NumPy arrays.</li>
<li> Understand important tensor attributes, such as shape, size, dim, dtype, and device.</li>

</ul>  
    
The unit also includes exercises to practice working with tensors, such as implementing one-hot encoding and other related tasks.

<a id='creation'></a> 
## Tensor Creation

The fundamental PyTorch data structure is a [**tensor**]((https://pytorch.org/docs/stable/tensors.html)) object, which represents a multidimensional array containing fixed-size items of a single data type, including integers, floats, and booleans. PyTorch tensors can be created in various ways. In the following, we present some initial examples and discuss fundamental properties of PyTorch tensors. More in-depth information about tensor attributes and data types can be found <a href='#attributes'>below</a>

We begin by importing the `torch` package, which includes data structures for multidimensional tensors and defines mathematical operations on these tensors. In the first example, we create a PyTorch tensor from a list using the `torch.tensor` function:

```python
import torch

x = [1,2,3,4]
print(f"x = {x}, type = {type(x)}")

x_torch = torch.tensor(x)
print(f"x_torch = {x_torch}, type = {type(x_torch)}")
```

In the next example, we create a PyTorch tensor from a NumPy array using the `torch.from_numpy` function. It is important to note that the resulting tensor and the NumPy array share the **same memory**. As a result, any modifications made to the tensor will also affect the array, and vice versa.

```python
import numpy as np

x_np = np.array([1,2,3,4])
print(f"x_np = {x_np}, type = {type(x_np)}")

x_torch = torch.from_numpy(x_np)
print(f"x_torch = {x_torch}, type = {type(x_torch)}")

x_torch[0] = 10
x_np[1] = 20
print(f"x_np = {x_np}")
print(f"x_torch = {x_torch}")
```

Moreover, PyTorch provides several functions for tensor creation. Here are some examples:

```python
shape = (3,)
print(f"Zeros Tensor: {torch.zeros(shape)}")
print(f"Ones Tensor: {torch.ones(shape)}")
print(f"Ones Tensor: {torch.ones(shape, dtype=int)}")
print(f"Random Tensor: {torch.rand(shape)}")

x = torch.tensor([[1,2,3], [4,5,6]])
print(f"Ones Tensor:\n {torch.ones_like(x)}")
torch.set_printoptions(precision=2)
print(f"Random Tensor:\n {torch.rand_like(x, dtype=torch.float)} ")
torch.set_printoptions(precision=4)
```

<a id='attributes'></a> 
## Tensor Attributes and Data Types

A PyTorch tensor comes with **attributes** that offer crucial information regarding the tensor's characteristics and configuration. Significant attributes of PyTorch tensors include:

* `shape`: Denotes the dimensions (size) of the tensor. 
* `size`: Provides the number of elements in each dimension of the tensor (equivalent to its shape).
* `dim`: Specifies the number of dimensions (axes) in the tensor.
* `dtype`: Indicates the data type stored in the tensor.
* `device`: Specifies whether the tensor resides on the CPU or a particular GPU device.
* `item`: Extracts Python value from a single-element tensor.

Here are some concrete examples:

```python
x = torch.ones(3,4)

print("Shape of tensor:", x.shape)
print("Size of tensor:", x.size())    #Equivalent to x.shape
print("Size in first dimension of tensor:", x.size(0)) 
print("Dimension of tensor:", x.dim()) 
print("Data type of tensor:", x.dtype) 
print("Device of tensor:", x.device) 
print("One element in tensor:", x[0,0]) 
print("Item of that element:", x[0,0].item()) 
```

<div class="alert alert-block alert-warning">
<strong>Note:</strong> The usage of the concept of <strong>dimension</strong> can be quite confusing at times. For instance, let us consider a real-valued tensor <code>x</code> with a shape of <code>(3,4)</code>. In mathematical terms, one might describe such a tensor as residing in a twelve-dimensional vector space, denoted as $\mathbb{R}^{3\times 4}$. However, when talking about the dimensions of a PyTorch tensor, as indicated by <code>x.dim</code> (which is two in our example), we are actually referring to what is commonly known as <strong>axes</strong>. In this context, <code>x.size(dim=None)</code> provides the dimension-wise size of the tensor <code>x</code>. To add to the confusion, the <code>size</code> attribute returns the number of elements for a <strong>NumPy array</strong>, while this count is given by the <code>numel</code> attribute for a <strong>PyTorch tensor</strong>. This terminology can indeed lead to confusion, as it depends on the context and the specific framework being used.
</div>

```python
x_np = np.ones((3,4))
print("x_np.size: ", x_np.size)

x_torch = torch.ones((3,4))
print("x_torch.size():", x_torch.size())
print("x_torch.numel():", x_torch.numel())
```

### Data Types and Conversion

Let us have a closer look at the attribute `dtype`.  PyTorch supports [twelve different data types](https://pytorch.org/docs/stable/tensor_attributes.html) for encoding integer, real-valued, and boolean values. Among these, the most important data types include:

* `torch.float32` or `torch.float`: This is the default PyTorch data type for real numbers.
* `torch.int64` or `torch.long`:  This is the default PyTorch data type for integer numbers.

Here are some examples:

```python
print("Data type:", torch.tensor([1,2,3]).dtype)
print("Data type:", torch.tensor([1,2,3], dtype=torch.float).dtype)
print("Data type:", torch.tensor([1,2,3], dtype=torch.int32).dtype)
print("Data type:", torch.tensor([1.,2,3]).dtype)
print("Data type:", torch.ones(1,2).dtype)
```

In PyTorch, one can cast a tensor of a given type into a tensor of another type in several ways. The method `.type()` accepts `dtype` as a parameter and returns a copy of the original tensor with the specified data type.

```python
x = torch.tensor([1.234,2.1,3.3])
print("x = ", x)
print("Data type of x:", x.dtype)
y = x.type(torch.int32)
print("y = ", y)
print("Data type of y:", y.dtype)
```

<div class="alert alert-block alert-warning">
<strong>Note:</strong>
The default data types employed for PyTorch tensors  (<code>torch.float32</code>, <code>torch.int64</code>) differ from the default data types used in NumPy arrays (<code>float64</code>, <code>int32</code>). When converting between NumPy arrays and PyTorch tensors, the data type is preserved, and they may even share the same underlying memory when stored on the CPU. This can lead to inconsistencies and confusion if data types are not explicitly specified. To explicitly convert the data type, you can use the `to` method. The following code cells provides some examples.
</div>

```python
x = np.array([1,2,3])
print("Data type for x:", x.dtype)
print("Data type for x:", torch.from_numpy(x).dtype)
y = torch.tensor([1,2,3])
print("Data type for y:", y.dtype)
print("Data type for y:", y.numpy().dtype)
z = torch.from_numpy(x).to(dtype=torch.int64)
print("Data type for z:", z.dtype)
```

### Device

The attribute `x.device` of a tensor `x` indicates the devise type (`cpu` , `cuda`) the tensor is allocated to. By default, tensors are created on the CPU. We will discuss <a href='#moving'>below</a> on how to move tensors between CPU and GPU.

```python
x = torch.tensor([1,2,3])
print(f"Device where the tensor is stored on: {x.device}")

device = ( "cuda" if torch.cuda.is_available() else "cpu" )
x = torch.tensor([1,2,3], device=device)
print(f"Device where the tensor is stored on: {x.device}")
```

<a id='operations'></a>
## Tensor Operations

PyTorch tensors offer a wide range of operations, including the following types:

* Elementwise operations, such as `torch.add()`, `torch.mul()`, `torch.div()`, ...
* Matrix and linear algebra operations, such as `torch.matmul()`, `torch.linalg.det()`, ...
* Reduction operations, such as `torch.sum()`, `torch.max()`, `torch.argmax()`, ...
* Indexing and reshaping operations, such as `torch.view()`, `torch.reshape()`, `torch.cat()`, ... 

For a comprehensive list of operations, we refer to the [Python documentation](https://pytorch.org/docs/stable/torch.html). In the following, we provide typical examples of some important operations. 

### Indexing and Slicing

In PyTorch, indexing and slicing are techniques used to access specific elements or subarray of a tensor. The syntax is similar to NumPy arrays. 

```python
x = torch.tensor([[1,2,3],[4,5,6]])
print("Tensor x:\n", x)
print("First row:", x[0])
print("Last column:", x[:, -1])
print("Element x[0,0]:", x[0,0])
print("Value of element x[0,0]:", x[0,0].item())
```

### Transposition,  Concatenation, Splitting

Common operations for manipulating tensors include:

* Transposition: This involves swapping the axes of a tensor.
* Concatenation: It entails combining two or more tensors along a specified dimension.
* Splitting: This divides a tensor into smaller tensors along a specified dimension.

Here are some examples:

```python
x = torch.tensor([[1,2], [3,4], [5,6]])
print("Tensor x:\n", x)
print("Transposition:\n", x.T)
print("Transposition:\n", x.transpose(0,1))

chunk_size = 1
y = x.split(chunk_size, dim=0)
print("Split tensor:\n", y)

z = y[1]
print(f"Concatenation:\n", torch.cat([z, z], dim=0))
print(f"Concatenation:", torch.cat([z, z], dim=1))
```

### Reshaping

Reshaping refers to the process of changing the shape of a tensor while preserving the total number of elements. It allows you to transform a tensor into a different shape without modifying the data it contains. Reshaping is a fundamental operation in deep learning and is frequently employed when dealing with neural networks, as distinct layers often necessitate input tensors in particular shapes. In the subsequent code cells, we present a few illustrative examples.

The `view()` function provides a tensor with a new shape while sharing the same data as the original tensor. On the other hand, the `reshape()` function serves a similar purpose but may or may not generate a copy of the original tensor, depending on whether the tensor is contiguous or non-contiguous (as explained <a href='#storage'>below</a>).

```python
x = torch.tensor([[1,2,3],[4,5,6]])
y = x.view(6)
z = x.reshape(6)
print("Shape of y:", y.shape)
print("Shape of z:", z.shape)
x[1,0] = 40
print("y:", y)
print("z:", z)
```

### Adding and Removing Dimensions

The `squeeze()` function produces a tensor where dimensions with a size of 1 are removed from the input tensor. Conversely, the `unsqueeze()` function generates a tensor by adding a dimension of size one at the specified position. This operation can also be accomplished by including dimensions with `None`. Note again that these operations do **not** copy the data but rather provide an alternate view of the existing data.

```python
x = torch.zeros([2,1,1,4])
print("Shape of x:", x.shape)
y = x.squeeze()
print("Shape of y:", y.shape)
z = y.unsqueeze(0)
print("Shape of z:", z.shape)
z2 = y[None,:,:,None]
print("Shape of z2:", z2.shape)
x[0,0,0,0] = 1
print("Tensor z after modifying x:\n", z)
```

### Permutation

The `permute()` function returns a view of the original tensor with its dimensions permuted.

```python
x = torch.zeros(2,3,4)
y = x.permute(2, 0, 1)
print("Shape of x:", x.shape)
print("Shape of y:", y.shape)
```

### Elementwise Matrix Operations


Elementwise matrix operations apply a specific operation to corresponding elements in two or more tensors, performing the operation independently on each pair of elements. Common examples of elementwise matrix operations include addition, subtraction, elementwise matrix multiplication (also known as the Hadamard product), exponentiation, logarithm, and elementwise comparison.

```python
x = torch.tensor([[0,1,2]], dtype=torch.float32)
y = torch.ones((1,3), dtype=torch.float32)*2
print("Addition:", x+y)
print("Addition:", torch.add(x,y))
print("Multiplication:", x*y)
print("Multiplication:", torch.mul(x,y))
print("Division:", x/y)
print("Exponention:", torch.exp(x))
print("Logarithm:", torch.log(x))
print("Comparison x<y:", x<y)
print("Comparison x==y:", x==y)
```

### Matrix and Linear Algebra Operations

Matrix and linear algebra operations are fundamental in machine learning, and PyTorch offers a comprehensive range of functions and methods for performing these operations efficiently. In the following code cells, we provide some illustrative examples, starting with matrix multiplication:

```python
x = torch.ones((1,3), dtype=torch.float32)
y = torch.tensor([[1,2],[3,4],[5,6]], dtype=torch.float32)
print("Matrix product:", torch.mm(x,y))
print("Matrix product:", torch.matmul(x,y))
print("Matrix product:", x @ y)
```

Here are some examples for computing the determinant, the inverse, and the eigenvalue decomposition for a square matrix.

```python
x = torch.tensor([[2,1],[1,2]], dtype=torch.float32)
print("Matrix determinant:", torch.det(x))
print("Matrix inversion:\n", torch.inverse(x))
print("Eigenvalues:", torch.linalg.eig(x)[0])
print("Eigenvectors:\n", torch.linalg.eig(x)[1])
```

### Reduction Operations

A reduction function is a function that combines multiple elements into a single result by reducing or aggregating them. In the following, we provide some illustrative examples and refer to the [PyTorch documentation](https://pytorch.org/docs/stable/tensors.html) for a comprehensive list:

```python
x = torch.tensor([[1,2,3],[6,5,4]], dtype=torch.float)
print("Global mean:", x.mean())
print("Standard deviation:", x.std())
print("Column-wise mean:", x.mean(0))
print("Column-wise standard deviation:", x.std(0))
print("Global sum:", x.sum())
print("Row-wise max:\n", x.max(1)[0])
print("Row-wise argmax:\n", x.max(1)[1])
print("Row-wise argmax:\n", x.argmax(1))
```

The function `permute` returns a view of the original tensor input with its dimensions permuted. In the following example, we generate a view (**not** a copy) by having axis 2 first and then axes 0 and 1

```python
x = torch.zeros(4,5,6)
print(f"Original tensor; shape = {x.shape}, stride = {x.stride()}")
y = x.permute(2,0,1)
print(f"Permuted tensor; shape = {y.shape}, stride = {y.stride()}")
```

<a id='storage'></a>
## Tensor Storage


The elements of a PyTorch tensor are organized within a one-dimensional array and stored in a contiguous block of memory, which is managed by a `storage` instance. A tensor is a view into this `storage` instance, using an offset and per-dimension strides. The attributes `.storage_offset` and `.stride` encode the relationship between tensor indices and storage indices.

```python
x = torch.tensor([[1,2,3], [4,5,6]])
print(f"x.shape: {x.shape}")
print(f"x.storage_offset = {x.storage_offset()}")
print(f"x.stride = {x.stride()}")
s = x.storage()
print(f"Storage:\n {s}")
```

Certain tensor operations do not change the values of the tensor but just the **view** on it. In other words, these operations do not create a new tensor, but just the way it is structured. As a result, changing the values of one view changes the values of the storage and all other views. This can be also tested using `.data_ptr()`, which return the address of the first element of the tensor.

```python
x = torch.tensor([[1,2], [3,4], [5,6]])
s = x.storage()
s[1]=20
y = x.T
print(f"Transposed tensor:\n {y}")
print(f"Same address of first element: {x.data_ptr() == y.data_ptr()}")
```

The internal storage layout of a tensor may vary depending on how it was originally defined. The method `.is_contiguous()` checks if a tensor is contiguous. In the case it is not, one can use `.contiguous()` to ensure that it becomes contiguous.

```python
x = torch.tensor([[1,2], [3,4], [5,6]])
y = x.T
z = y.contiguous()
print("x.stride:", x.stride())
print("Contiguous:", x.is_contiguous())
print("y.stride:", y.stride())
print("Contiguous:", y.is_contiguous())
print("z.stride:", z.stride())
print("Contiguous:", z.is_contiguous())

x[0,1]=20
print("y:\n", y)
print("z:\n", z)
print(f"Same address of first element: {y.data_ptr() == z.data_ptr()}")
```

The function `clone()` returns a copy of the input, thus creating a new storage instance.

```python
x = torch.tensor([[1,2,3], [4,5,6]])
y = x.view(6)
z = x.view(6).clone()
print(f"Same address of first element: {x.data_ptr() == y.data_ptr()}")
print(f"Same address of first element: {x.data_ptr() == z.data_ptr()}")
```

<a id='moving'></a>
## Moving Tensors Between Devices and Types

By default, PyTorch stores a tensor on the CPU. One can transfer a tensor to a GPU to accelerate computations. In addition to the data type (`dtype`), a tensor also has the `device` attribute, which indicates its placement. To move a tensor between different processors, one can use the `.to` method. This method can also be combined with type conversion using the `dtype` attribute. 

```python
x = torch.tensor((1.1,2.0,3.6))
y = x.to(torch.int32)
print(f"Data type: {x.dtype}; tensor: {x}")
print(f"Data type: {y.dtype}; tensor: {y}")

print(f"Device (default): {x.device}")
device = ("cuda" if torch.cuda.is_available() else "cpu")
z = x.to(dtype=torch.float64, device=device)
print(f"Device (after conversion): {z.device}; data type = {z.dtype}")
```

The function `.item()` moves a one-element tensor to CPU and converts it into a standard Python number. More general, the function `.tolist()` converts a tensor into a (nested) list.

```python
x = torch.tensor([4], device=device, dtype=torch.float32)
print(x)
print(x.item())

x = torch.tensor([[1,2,3],[4,5,6]], device=device, dtype=torch.float32)
print(x)
y = x.tolist()
print(y)
```

<a id='serializing'></a>
##  Serializing Tensors

The term **serialization** refers the process of converting an object structure into a format that can be stored. The opposite operation of reconstructing the object structure from that format is referred to as **deserialization**. A tensor can be stored by using the `pickle` module, which provides a protocol for serializing and deserializing Python object structures. The following examples shows how one can save and load a tensor.

```python
import os
file_dir = "output"
file_name = "tensor_saved.t"
file_dir_name = os.path.join(file_dir, file_name)

x =  torch.tensor([[1,2,3], [4,5,6]])
torch.save(x, file_dir_name)

x_load =  torch.load(file_dir_name)
print(x_load)
```

<div class="alert alert-block alert-warning">
<strong>Note:</strong>
Various operating systems employ distinct conventions for path names. For instance, Windows utilizes backslashes (<code>\</code>) as path separators, whereas Linux environments employ forward slashes (<code>/</code>). To mitigate compatibility issues when specifying paths and file names, one approach is to utilize the <code>os</code> module, which provides functions for interacting with the underlying operating system.
</div>

As an alternative, to allow for interoperability with other systems, one can use [Hierarchical Data Format version 5](https://www.hdfgroup.org/solutions/hdf5/) (HDF5), which is a general purpose library and file format for storing large amounts of complex and heterogeneous data. Python supports HDF5 through the `h5py` library.

<strong>Exercise 1: One-Hot Encoding</strong><br>
One-hot encoding is a common technique used in machine learning to represent categorical data as binary vectors. Given a finite set $\mathcal{A}=\{\alpha_0,\ldots,\alpha_{N-1}\}$ of $N$ distinct <strong>labels</strong> (also called <strong>alphabet</strong> of size $K$), then one-hot encoding assigns to each label $\alpha_k$ for $k\in\{0,...,K-1\}$ a distinct binary vector $v_k\in\mathbb{R}^K$ having exactly one non-zero element of value $1$. For example, given the alphabet <code>{'A','B','C'}</code> consisting of three letters, one could encode <code>'A'</code> by <code>[1,0,0]</code>, <code>'B'</code> by <code>[0,1,0]</code>, and <code>'C'</code> by <code>[0,0,1]</code> (here, the vectors $v_k$ are given as row vectors). From this assignment, one obtains a one-hot encoding of any finite sequence over the alphabet $\mathcal{A}$. For example, the label sequence <code>('C','A','B','A')</code> is represented by the $4\times 3$ matrix <code>[[0,0,1], [1,0,0], [0,1,0], [1,0,0]]</code>. Write a function <code>compute_one_hot_encoding</code> with the following specifications:
    
<ul>
    <li>Input: An alphabet and a sequence over the alphabet (both given as list)</li>
    <li>Output: One-hot encoding of the sequence</li>
    <li>Make sure that the alphabet only contains unique labels (using <code>set()</code></li>
    <li>Sort the elements of the alphabet alphanumerically (using <code>sorted()</code>)</li>
    <li>Make sure that all elements in the sequence are in the alphabet (otherwise return <code>None</code>)</li>    
</ul> 
    
Evaluate your function for the following inputs:    
    
<ul>
    <li>Input: Alphabet $\mathcal{A}=$<code>('B','A','C')</code> and sequence <code> ('A','A','B','C','A','A','B')</code></li>
    <li>Input: Alphabet $\mathcal{A}=$<code>('A','B','C')</code> and sequence <code> ('A','B','C','D')</code></li>    
    <li>Input: Alphabet $\mathcal{A}=$<code>('hi','a','*')</code> and sequence <code> ('*','hi','a','a','*','*')</code></li>
</ul>
